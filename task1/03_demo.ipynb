```python
from transformers import pipeline

MODEL_PATH = "YOUR_USERNAME/mountain-ner-distilbert"

ner = pipeline(
    "token-classification",
    model=MODEL_PATH,
    tokenizer=MODEL_PATH,
    aggregation_strategy="simple",
)
```

```python
text = "Mount Everest and Lhotse were visible, while the word summit appeared in the report."
predictions = ner(text)
predictions
```

```python
import pandas as pd

result = pd.DataFrame([
    {
        "entity": item["word"],
        "label": item["entity_group"],
        "confidence": round(float(item["score"]), 4),
        "start": item["start"],
        "end": item["end"],
    }
    for item in predictions
])
result
```

```python
from IPython.display import HTML, display

def highlight_entities(text, predictions):
    pieces = []
    cursor = 0
    for item in sorted(predictions, key=lambda x: x["start"]):
        pieces.append(text[cursor:item["start"]])
        pieces.append(
            f'<mark title="{item["score"]:.3f}">'
            f'{text[item["start"]:item["end"]]}</mark>'
        )
        cursor = item["end"]
    pieces.append(text[cursor:])
    return "".join(pieces)

display(HTML(highlight_entities(text, predictions)))
```
